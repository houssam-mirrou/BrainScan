## 1. The Artificial Neuron

An artificial neuron receives numerical inputs, multiplies
them by weights, adds a bias, and applies an activation function.

Mathematically:

z = w1*x1 + w2*x2 + ... + wn*xn + b

a = f(z)

Where:
- x: inputs
- w: weights
- b: bias
- z: weighted sum
- f: activation function
- a: neuron output

Weights and biases are learnable parameters that are
adjusted during model training.


```python

import numpy as np

# Inputs
x = np.array([0.2, 0.6, 0.9])

# Weights
w = np.array([0.5, -0.4, 0.8])

# Bias
b = -0.1

# Weighted sum
z = np.dot(x, w) + b

# ReLU activation
a = max(0, z)

print("Inputs:", x)
print("Weights:", w)
print("Bias:", b)
print("Weighted sum:", round(z, 4))
print("Neuron output:", round(a, 4))
```

```python
x = np.array([0.2, 0.6, 0.9])
b = -0.1

weight_sets = [
    [0.5, -0.4, 0.8],
    [0.1, 0.1, 0.1],
    [-0.5, -0.4, -0.8]
]

for weights in weight_sets:

    w = np.array(weights)

    z = np.dot(x, w) + b
    output = max(0, z)

    print(
        f"Weights: {weights} | "
        f"Weighted sum: {z:.3f} | "
        f"Output: {output:.3f}"
    )
```


## 2. Activation Functions

Activation functions introduce nonlinearity into neural networks.

Without nonlinear activation functions, stacking multiple
linear layers is mathematically equivalent to a single
linear transformation.

### ReLU

ReLU(x) = max(0, x)

- Returns 0 for negative values.
- Returns x for positive values.
- Commonly used in hidden layers.

### Sigmoid

Sigmoid(x) = 1 / (1 + exp(-x))

- Produces values between 0 and 1.
- Commonly used for binary classification outputs.

### Softmax

Softmax converts a vector of scores into values that
sum to 1.

- Commonly used for multiclass classification.
- We will use Softmax for our four output classes:
  glioma, meningioma, notumor, and pituitary.


```python
import numpy as np

def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.array([-3, -1, 0, 1, 3])

print("Inputs:", x)
print("ReLU:", relu(x))
print("Sigmoid:", np.round(sigmoid(x), 3))
```

```python
def softmax(logits):

    # Subtract the maximum for numerical stability
    shifted = logits - np.max(logits)

    exponentials = np.exp(shifted)

    return exponentials / np.sum(exponentials)


logits = np.array([2.0, 1.0, 0.0, -1.0])

probabilities = softmax(logits)

class_names = [
    "glioma",
    "meningioma",
    "notumor",
    "pituitary"
]

for name, probability in zip(class_names, probabilities):
    print(f"{name}: {probability:.2%}")

print("\nSum:", probabilities.sum())
print("Predicted class:", class_names[np.argmax(probabilities)])
```


## 3. Neural Network Layers

A neural network consists of neurons organized into layers.

### Types of layers

- Input layer: receives the input features.
- Hidden layers: apply learned weights, biases, and
  activation functions to transform the inputs.
- Output layer: produces the final prediction.

### Fully Connected Layer

In a fully connected (Dense) layer, every neuron receives
the outputs from all neurons in the previous layer.

A layer can be represented mathematically as:

Z = XW + b

A = activation(Z)

Where:
- X: input values
- W: weight matrix
- b: bias vector
- Z: weighted sums
- A: layer outputs

Each neuron has its own weights and bias.

The number of output values equals the number of
neurons in the layer.


```python
import numpy as np

# Three input features
x = np.array([1.0, 0.5, 0.2])

# Each column contains one neuron's weights
W = np.array([
    [0.2, -0.5],
    [0.4,  0.1],
    [-0.3, 0.8]
])

# One bias per neuron
b = np.array([0.1, -0.2])

# Forward calculation
z = np.dot(x, W) + b

# ReLU activation
a = np.maximum(0, z)

print("Input shape:", x.shape)
print("Weight matrix shape:", W.shape)
print("Bias shape:", b.shape)

print("\nWeighted sums:", z)
print("Layer outputs:", a)
```

## Forward Propagation

A neural network passes input data through layers.

For a Dense layer:

Z = XW + b

A = activation(Z)

The output of one layer becomes the input to the next.

## Loss Function

The loss function measures the difference between
the model's prediction and the correct answer.

For multiclass classification, we use categorical
cross-entropy or sparse categorical cross-entropy.

## Gradient Descent and Backpropagation

Gradient descent updates model parameters to reduce loss.

Backpropagation uses the chain rule to calculate
how the loss changes with respect to each parameter.

The basic update rule is:

weight = weight - learning_rate * gradient

The learning rate determines the size of each update.


```python

import numpy as np

rng = np.random.default_rng(42)

# Three input features
x = np.array([0.2, 0.6, 0.9])

# Hidden layer: 3 inputs -> 4 neurons
W1 = rng.normal(size=(3, 4))
b1 = np.zeros(4)

# Output layer: 4 neurons -> 4 classes
W2 = rng.normal(size=(4, 4))
b2 = np.zeros(4)

# Forward propagation
z1 = x @ W1 + b1
a1 = np.maximum(0, z1)

z2 = a1 @ W2 + b2

# Softmax
exp_values = np.exp(z2 - np.max(z2))
probabilities = exp_values / exp_values.sum()

# Example true class: notumor = 2
true_class = 2

loss = -np.log(
    np.clip(probabilities[true_class], 1e-7, 1)
)

print("Hidden layer output:", a1)
print("Class probabilities:", probabilities)
print("Predicted class ID:", np.argmax(probabilities))
print("Cross-entropy loss:", loss)
```
