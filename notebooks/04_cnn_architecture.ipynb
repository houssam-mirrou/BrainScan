
# 04 — CNN Architecture

## Objective

Build a convolutional neural network to classify brain
images into four categories.

Our architecture uses:

- Conv2D: learns spatial filters and feature maps.
- ReLU: introduces nonlinearity.
- MaxPooling2D: reduces spatial dimensions.
- GlobalAveragePooling2D: converts feature maps into
  a compact feature vector.
- Dense: combines the extracted features.
- Dropout: randomly disables some activations during
  training to help reduce overfitting.
- Softmax: produces four class probability scores.
